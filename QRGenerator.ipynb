```python
import qrcode
from PIL import Image, ImageDraw

# === CONFIGURACIÓN ===
url = "www.google.com"  # Reemplaza con tu URL
output_path = "output/qr_impronta.png"

# === GENERAR QR ===
qr = qrcode.QRCode(
    version=5,
    error_correction=qrcode.constants.ERROR_CORRECT_H,  # Alta tolerancia para insertar logo
    box_size=10,
    border=4,
)
qr.add_data(url)
qr.make(fit=True)

# === CREAR IMAGEN QR CON COLORES PERSONALIZADOS ===
qr_img = qr.make_image(fill_color="#007A5E", back_color="white").convert("RGB")  # Verde institucional

# === GUARDAR RESULTADO ===
qr_img.save(output_path)
print(f"✅ QR empresarial generado y guardado como '{output_path}'")
```
